### Morning Time ANN Model creation

Time --> 6:30 am to 11 pm 

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('../Datasets/traffic_light_morning_time.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Remove time_of_day column
df.drop(columns=['time_of_day','priority_order'],axis=1,inplace=True)
df.head()
```

```python
X=df.drop(columns=["green_N","green_S","green_E","green_W"])
Y=df[["green_N","green_S","green_E","green_W"]]
```

```python
X
```

```python
Y
```

```python
# split test and training data
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=.2,random_state=15)
```

```python
# stanrddization data for independent values
from sklearn.preprocessing import StandardScaler

morning_time_scalar_independent=StandardScaler()
X_train=morning_time_scalar_independent.fit_transform(X_train)
X_test=morning_time_scalar_independent.transform(X_test)
```

```python
# stanrddization data for independent values
from sklearn.preprocessing import StandardScaler

morning_time_scalar_dependent=StandardScaler()
Y_train=morning_time_scalar_dependent.fit_transform(Y_train)
Y_test=morning_time_scalar_dependent.transform(Y_test)
```

```python
# save StandardScaler models

import pickle

with open("../models/morning_time_models/morning_time_scaler_independent.pkl","wb") as file:
	pickle.dump(morning_time_scalar_independent,file)

with open("../models/morning_time_models/morning_time_scaler_dependent.pkl","wb") as file:
	pickle.dump(morning_time_scalar_dependent,file)
```

### ANN Training

```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
# Model creation
morning_time_model = Sequential([
    Dense(64, input_dim=X_train.shape[1], activation='relu'),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(4, activation='linear')
])
```

```python
morning_time_model.summary()
```

```python
optimiser=tensorflow.keras.optimizers.Adam(learning_rate=.1)
loss=tensorflow.keras.losses.mse
```

```python
# set optimisers , loss functions
morning_time_model.compile(optimizer=optimiser,loss=loss,metrics=['mae'])
```

```python
# set up early stoping 
early_stopping_call_back=EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
```

```python
# set up log files

log_dir="../logs/morning_time_log/fit/"+datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorBoard_callback=TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
morning_time_model.fit(
    X_train,
    Y_train,
    epochs=1000,
    validation_data=(X_test, Y_test),
    callbacks=[tensorBoard_callback, early_stopping_call_back]
)
```

```python
morning_time_model.metrics
```

```python
results = morning_time_model.evaluate(X_test, Y_test, return_dict=True)
print(results)
```
